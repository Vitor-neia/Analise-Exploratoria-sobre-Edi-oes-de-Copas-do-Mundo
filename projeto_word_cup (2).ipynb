```python
df['media_gols'] = df['goals_scored'] / df['games']  # Média de gols
df
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
plt.figure(figsize=(8,6))
plt.plot(df['year'], df['media_gols'], marker='o', color='darkblue')
idx_max = df['media_gols'].idxmax()
ano_max = df.loc[idx_max, 'year']
valor_max = df.loc[idx_max, 'media_gols']
plt.annotate(
    f'Maior Média\n{valor_max:.2f} ({ano_max})',
    xy=(ano_max, valor_max),
    xytext=(ano_max + 4, valor_max - 0.6),
    arrowprops=dict(arrowstyle='->', color='red'),
    fontsize=9,
    color='black'
)
idx_min = df['media_gols'].idxmin()
ano_min = df.loc[idx_min, 'year']
valor_min = df.loc[idx_min, 'media_gols']
plt.annotate(
    f'Menor Média\n{valor_min:.2f} ({ano_min})',
    xy=(ano_min, valor_min),
    xytext=(ano_min + 9, valor_min + 0.7),
    arrowprops=dict(arrowstyle='->', color='red'),
    fontsize=9,
    color='black'
)


plt.title('Média de gols por Copa do Mundo', fontweight='bold')
plt.xlabel('Ano', fontweight='bold')
plt.ylabel('Média de gols', fontweight='bold')
plt.xticks(df['year'], rotation=45)

plt.ylim(2, 6)
plt.yticks(np.arange(0, 6.5, 0.5))
plt.tight_layout()
plt.show()
```

Evolução da Média de Gols em Copas do Mundo
Pico de gols: O recorde de média de gols ocorreu em 1954, com 5,38 gols por partida.

Menor média: A menor média foi registrada em 1990, com 2,21 gols por jogo.

Tendência histórica: De modo geral, as médias de gols eram mais altas antes dos anos 2000. Isso reflete a crescente profissionalização, o rigor tático e o foco defensivo que o futebol assumiu nas últimas décadas.

```python
df['media_publico'] =  df['attendance'] / df['games']  # Média de público
df
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
plt.figure(figsize=(8,6))
plt.plot(df['year'], df['media_publico'], marker='o', linewidth=2.5, color='#e63946')
plt.axhline(df['media_publico'].mean(), color='gray', linestyle='--', label='Média geral')
idx_max =  df['media_publico'].idxmax()
plt.annotate('Recorde\nEUA 1994',
             xy=(df['year'][idx_max], df['media_publico'][idx_max]),
             xytext=(1996, 65000),
             arrowprops=dict(arrowstyle='->'))
idx_min = df['media_publico'].idxmin()
plt.annotate('Mínimo\nItália 1934',
             xy=(df['year'][idx_min], df['media_publico'][idx_min]),
             xytext=(1936, 27500),
             arrowprops=dict(arrowstyle='->', color='black', lw=1.2))

plt.title('Média de público por Copa do Mundo', fontweight='bold')
plt.ylabel('Média de público', fontweight='bold')
plt.xlabel('Ano', fontweight='bold')
plt.xticks(df['year'], rotation=45)

plt.legend()
plt.tight_layout()
plt.show()
```

Recordes de Público em Copas do Mundo
Recorde de público: A maior média de público por partida foi registrada na Copa de 1994, nos Estados Unidos.

Mínimo de público: A menor média de público por jogo ocorreu na Copa de 1934, realizada na Itália.

```python
def desempenho_sede(row):
    host = row['host']
    hosts = [h.strip() for h in host.split(',')]

    for h in hosts:
        if h == row['winner']:
            return 'Campeão'
        elif h == row['second']:
            return 'Vice'
        elif h == row['third']:
            return '3º lugar'
        elif h == row['fourth']:
            return '4º lugar'
    return 'Não chegou ao pódio'

df['desempenho_sede'] = df.apply(desempenho_sede, axis=1)

resultado_sede = df[['year', 'host', 'desempenho_sede']].copy()
resultado_sede.columns = ['Ano', 'Sede', 'Desempenho']
print(resultado_sede.to_string(index=False))

print()

contagem = df['desempenho_sede'].value_counts()
print(contagem)

ordem = ['Campeão', 'Vice', '3º lugar', '4º lugar', 'Não chegou ao pódio']
cores = ['gold', 'silver', '#cd7f32', '#6baed6', '#bdbdbd']

contagem_ordenada = contagem.reindex(ordem, fill_value=0)

fig, ax = plt.subplots(figsize=(9, 5))

bars = ax.bar(contagem_ordenada.index, contagem_ordenada.values, color=cores, edgecolor='white', linewidth=1.2)

for bar in bars:
    height = bar.get_height()
    if height > 0:
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            height + 0.05,
            str(int(height)),
            ha='center', va='bottom', fontweight='bold', fontsize=11
        )

ax.set_title('Desempenho do País-Sede na Copa do Mundo', fontweight='bold', fontsize=13)
ax.set_xlabel('Desempenho', fontweight='bold')
ax.set_ylabel('Número de Copas', fontweight='bold')
ax.set_ylim(0, contagem_ordenada.max() + 1.5)
ax.yaxis.set_major_locator(plt.MaxNLocator(integer=True))
plt.grid(False)
plt.style.use('seaborn-v0_8-whitegrid')
plt.yticks([])
plt.tight_layout()
plt.show()
```

O Fator Casa nas Copas do Mundo
As seleções anfitriãs costumam ter uma certa vantagem por jogar em casa, especialmente as seleções tradicionais. Exemplos disso são Uruguai, Itália, Inglaterra, Alemanha, Argentina e França; todas essas seleções conquistaram o torneio jogando diante de sua própria torcida.

```python
plt.style.use('seaborn-v0_8-whitegrid')

fig, ax = plt.subplots(2, 3, figsize=(18, 12))
fig.suptitle('Análise Exploratória sobre Edições de Copas do Mundo',
             fontweight='bold', fontsize=18, y=0.98)


df['winner'].value_counts().sort_values(ascending=True).plot(
    kind='barh', color='gold', ax=ax[0, 0]
)
ax[0, 0].set_title('Maiores Campeões do Mundo', fontweight='bold')
ax[0, 0].set_xlabel('Títulos', fontweight='bold')
ax[0, 0].set_ylabel('Seleções', fontweight='bold')
ax[0, 0].grid(False)


df['second'].value_counts().sort_values(ascending=True).plot(
    kind='barh', color='#C0C0C0', ax=ax[0, 1]
)
ax[0, 1].set_title('Seleções com mais vices', fontweight='bold')
ax[0, 1].set_xlabel('Vices', fontweight='bold')
ax[0, 1].set_ylabel('Seleções', fontweight='bold')
ax[0, 1].set_xticks(np.arange(1, 5, 1))
ax[0, 1].grid(False)


ax[0, 2].plot(df['year'], df['media_gols'], marker='o', color='darkblue')

idx_max = df['media_gols'].idxmax()
ano_max = df.loc[idx_max, 'year']
valor_max = df.loc[idx_max, 'media_gols']

ax[0, 2].annotate(f'Maior: {valor_max:.2f} ({ano_max})',
                  xy=(ano_max, valor_max),
                  xytext=(ano_max + 8, valor_max - 0.5),
                  arrowprops=dict(arrowstyle='->', color='red'), fontsize=9)

idx_min = df['media_gols'].idxmin()
ano_min = df.loc[idx_min, 'year']
valor_min = df.loc[idx_min, 'media_gols']

ax[0, 2].annotate(f'Menor: {valor_min:.2f} ({ano_min})',
                  xy=(ano_min, valor_min),
                  xytext=(ano_min + 8, valor_min + 0.5),
                  arrowprops=dict(arrowstyle='->', color='red'), fontsize=9)

ax[0, 2].set_title('Média de gols por Copa', fontweight='bold')
ax[0, 2].set_xlabel('Ano', fontweight='bold')
ax[0, 2].set_ylabel('Média', fontweight='bold')
ax[0, 2].set_ylim(2, 6)
ax[0, 2].tick_params(axis='x', rotation=45)


ax[1, 0].plot(df['year'], df['media_publico'], marker='o', linewidth=2.5, color='#e63946')
ax[1, 0].axhline(df['media_publico'].mean(), color='gray', linestyle='--', label='Média geral')
ax[1, 0].set_title('Média de público por Copa', fontweight='bold')
ax[1, 0].set_xlabel('Ano')
ax[1, 0].set_ylabel('Público médio', fontweight='bold')
ax[1, 0].tick_params(axis='x', rotation=45)
ax[1, 0].legend()

idx_max = df['media_publico'].idxmax()
ax[1, 0].annotate('Recorde\nEUA 1994', xy=(df['year'][idx_max], df['media_publico'][idx_max]),
                  xytext=(1970, 60000), arrowprops=dict(arrowstyle='->'))

idx_min = df['media_publico'].idxmin()
ax[1, 0].annotate('Mínimo\nÍtalia 1934', xy=(df['year'][idx_min], df['media_publico'][idx_min]),
                  xytext=(1938, 30000), arrowprops=dict(arrowstyle='->'))


contagem = df['desempenho_sede'].value_counts()
ordem_original = ['Campeão', 'Vice', '3º lugar', '4º lugar', 'Não chegou ao pódio']
ordem_display  = ['Campeão', 'Vice', '3º lugar', '4º lugar', 'Não chegou\nao pódio']
cores = ['gold', 'silver', '#cd7f32', '#6baed6', '#bdbdbd']

contagem_ordenada = contagem.reindex(ordem_original, fill_value=0)
contagem_ordenada.index = ordem_display

bars = ax[1, 1].bar(contagem_ordenada.index, contagem_ordenada.values, color=cores, edgecolor='white')

for bar in bars:
    height = bar.get_height()
    ax[1, 1].text(bar.get_x() + bar.get_width() / 2, height + 0.2, str(int(height)),
                  ha='center', va='bottom', fontweight='bold')

ax[1, 1].set_title('Desempenho do País-Sede', fontweight='bold')
ax[1, 1].set_yticks([])
ax[1, 1].grid(False)


ax[1, 2].set_visible(False)




plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

Esta análise revelou que o Brasil domina o quadro de títulos, apesar de estar há 24 anos sem vencer (a última conquista foi em 2002). O "fator casa" mostra-se historicamente decisivo, dado que diversas seleções tradicionais conquistaram o torneio jogando diante de sua própria torcida.

Observou-se também que a média de gols caiu significativamente após os anos 2000, um reflexo direto da crescente profissionalização e do rigor tático imposto ao jogo moderno. Por fim, a Copa de 1994, sediada nos Estados Unidos, permanece como o maior evento em termos de público na história do torneio.

Pontos principais consolidados:
Domínio Brasileiro: Pentacampeonato consolidado.

Vantagem de Campo: O apoio da torcida e a familiaridade com o ambiente como diferencial competitivo.

Evolução Tática: O impacto do sistema defensivo na redução da média de gols.

Recorde de Público: O sucesso comercial e estrutural da Copa de 1994.
